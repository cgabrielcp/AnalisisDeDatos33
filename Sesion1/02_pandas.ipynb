# Acceder a API

```python
import pandas as pd
import requests

url = "https://api.open-meteo.com/v1/forecast"

params = {
    "latitude": -12.04,
    "longitude": -77.04,
    "hourly": "temperature_2m",
    "timezone": "America/Lima",
}

data = requests.get(url, params= params).json() # Pedir la data

df = pd.DataFrame(data) # Convertir

df.head() # usar
```

```python
url = "https://www.alphavantage.co/query"

params = {
    "function": "TIME_SERIES_DAILY",
    "symbol": "IBM",
    "apikey": "demo",        # reemplazar por tu clave
}

data = requests.get(url, params=params).json()
df = pd.DataFrame(data["Time Series (Daily)"]).T

df.head()
```
