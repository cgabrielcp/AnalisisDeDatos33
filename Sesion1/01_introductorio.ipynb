# listas 

```python
# Lista de enteros
my_list_1 = [3, 5, 2]
print(my_list_1)

# Lista vacia
my_list_2 = []
print(my_list_2)

# Lista Mixta
my_list_3 = [3, 'Hello', 44.5]
print(my_list_3)

# Lista Anidada
my_list = ['apple', [3, ["a", 2, 2.5], 8], ['x']]
print(my_list)
```

```python
my_list.insert(1, 30)

my_list
```

```python
# Lista simple
my_list = ['x', 'l', 't', 'a']

print(my_list[3])
```

```python
#lista Anidad
n_list = ["Happy", [6, 2, 1, 8]]

print(n_list[0][2])

print(n_list[1][2])
```

```python
# Cambiar o actualizar un elemento de una lista

print(my_list)

my_list[0]="r"

print(my_list)
```

```python
# Podemos Iterar la lista
frutas = ['apple', 'banan', 'mango']

for fruta in frutas:
    print('I like ', fruta)
```
